# Data Visualization with Matplotlib
This notebook demonstrates various basic plot types using the `matplotlib` library, including bar charts, scatter plots, line plots, and pie charts.

```python
# Import the Pyplot interface from Matplotlib
import matplotlib.pyplot as plt
# Import the NumPy library for numerical operations
import numpy as np
```

## Define Sample Data
Here, we initialize sample numerical lists `x`, `y`, `z`, and `j` to use across our plotting examples.

```python
# Define the x-axis values for the first dataset
x = [2, 4, 6, 8]
# Define the y-axis values for the first dataset
y = [1, 3, 5, 7]

# Define the x-axis values for the second dataset
z = [7, 5, 3, 1]
# Define the y-axis values for the second dataset
j = [8, 6, 4, 2]
```

## 1. Bar Chart
Below is a simple bar chart representing the relationship between dataset `x` and dataset `y`.

```python
# Create a bar chart with skyblue colored bars
plt.bar(x, y, color = 'skyblue')
# Add a title to the bar chart
plt.title("A TEST TITLE")
# Label the x-axis of the plot
plt.xlabel("A TEST LABLE")
# Label the y-axis of the plot
plt.ylabel("A TEST LABEL")
# Render and display the generated bar chart
plt.show()
```

## 2. Scatter Plot
This section displays a scatter plot mapping two different series: `(x, y)` and `(z, j)` on a grid.

```python
# Create a scatter plot for the first dataset
plt.scatter(x, y)
# Create a scatter plot for the second dataset
plt.scatter(z, j)
# Set the title of the scatter plot
plt.title("A TEST TITLE")
# Set the label for the x-axis
plt.xlabel("A TEST LABLE")
# Set the label for the y-axis
plt.ylabel("A TEST LABEL")
# Turn on the background grid lines
plt.grid()
# Display the final scatter plot
plt.show()
```

## 3. Custom Line Plots
Here, we visualize the coordinate relationships as dashed line plots with custom markers (`*`) and unique colors.

```python
# Plot the first dataset as a dashed pink line with star markers
plt.plot(x, y, "*", ls = "dashed", color = 'pink')
# Plot the second dataset as a dashed purple line with star markers
plt.plot(z, j, "*", ls = "dashed", color = 'purple')
# Apply the title to the plot
plt.title("A TEST TITLE")
# Apply the label for the x-axis
plt.xlabel("A TEST LABLE")
# Apply the label for the y-axis
plt.ylabel("A TEST LABEL")
# Enable grid lines on the plot
plt.grid()
# Render and output the plot
plt.show()
```

## 4. Pie Chart
Finally, we present a pie chart using values from `x`, complete with customized colors, a starting angle of 90 degrees, and a legend.

```python
# Define the color palette for each pie chart slice
mycolors = ['blue', 'yellow', 'pink', 'orange']
# Draw a pie chart of x starting at an angle of 90 degrees
plt.pie(x, colors = mycolors, startangle=90)
# Draw a legend with a custom title and data labels
plt.legend(title = "TEST", labels = y)
# Set the overall title of the pie chart
plt.title("TEST")
```
