### Tiger Species Pie Chart
This cell visualizes the population estimates of wild tiger subspecies. It utilizes `numpy` to store the numbers and labels, and `matplotlib` to render a styled pie chart using a custom color mapping and a legend.

### Tiger Population Distribution in the Wild
This notebook visualizes the distribution of various tiger subspecies remaining in the wild using a pie chart with a customized color palette and a clear explanatory legend.

````python
import matplotlib.pyplot as plt
import numpy as np

# Define the estimated wild population for each tiger subspecies
Number_of_tigers = np.array([60, 350, 400, 2500, 500, 550])

# Define labels corresponding to each subspecies
types = ["south china", "indo_chinese", "siberian", "malayan", "bengal", "sumatran"]

# Color mapping for each wedge of the pie chart
mycolors = ["purple", "blue", "green", "yellow", "orange", "red"]

# Plot the pie chart with a custom starting angle
plt.pie(Number_of_tigers, colors=mycolors, startangle=90)

# Place a legend showing the corresponding labels
plt.legend(title="Tigers:", labels=types)

# Set the chart title
plt.title("Tigers in the wild")
````

### Tiger Weight Comparison over Time
This section visualizes and compares the growth (weight in kilograms) of two tigers, T.J. and Mattew, at different ages (measured in weeks) using a grouped bar chart.

````python
import matplotlib.pyplot as plt
import numpy as np

# Ages in weeks when weights were recorded
x = np.array([0, 6, 10])

# Weights in kilograms for tiger 'T.J'
y = np.array([1.5, 5, 6])

# Weights in kilograms for tiger 'Mattew'
z = np.array([1.5, 7, 9])

# Create an array representing the position indices on the x-axis
x_axis = np.arange(len(x))

# Plot two sets of bars side-by-side with an offset of 0.2
plt.bar(x_axis - 0.2, y, 0.4, label="T.J", color="purple")
plt.bar(x_axis + 0.2, z, 0.4, label="Mattew", color="pink")

# Match the x-axis indices with the actual age values
plt.xticks(x_axis, x)

# Add descriptive labels, a title, and display the legend
plt.xlabel("AGE IN WEEKS")
plt.ylabel("WEIGHT IN KILOGRAM")
plt.title("TIGERS WEIGHT")
plt.legend()

plt.show()
````

### Dotted Development Trajectory
This plot uses customized star markers and dashed connection lines to visualize the comparative growth patterns of two subjects, with an added grid system for easy estimation.

````python
# Import the matplotlib plotting library
import matplotlib.pyplot as plt
# Import numpy library for numerical data manipulation
  import numpy as np

# Create array of age values for the first subject
x = np.array([0, 6, 10, 11])
# Create array of weight values for the first subject
y = np.array([1.5, 5, 6, 5.5])

# Create array of age values for the second subject
z = np.array([2, 8, 13, 15])
# Create array of weight values for the second subject
j = np.array([2, 7, 9, 10.5])

# Plot the first subject's growth with star markers, dashed lines, and red color
plt.plot(x, y, "*", ls = "dashed", color = "red")
# Plot the second subject's growth with star markers, dashed lines, and orange color
plt.plot(z, j, "*", ls = "dashed", color = "orange")
# Label the x-axis to represent age in weeks
plt.xlabel("AGE IN WEEKS")
# Label the y-axis to represent weight in kilograms
plt.ylabel("WEIGHT IN KILOGRAM")
# Enable the background grid for easier value estimation
plt.grid()
# Render the finalized plot to the screen
plt.show()
````

### Mountain Height Mean Analysis

### Summary Weight Metrics Calculation
This code produces random object weights, assigns them into a DataFrame structure, and evaluates basic mathematical properties like average (mean), minimum, and maximum.

````python
# Import the pandas library for structured dataframes
import pandas as pd
# Import the numpy library for array math operations
import numpy as np

# Produce an array containing 10 randomized integer weights in the [10, 50) range
weights = np.random.randint(10, 50, 10)
# Store the generated weight elements into a DataFrame
df = pd.DataFrame({'weight':weights})

# Solve for the mathematical mean weight of the column sample
mean_weights = df['weight'].mean()
# Isolate the highest single value from the weight collection
max_weights = df['weight'].max()
# Isolate the lowest single value from the weight collection
min_weights = df['weight'].min()

# Display computed mean metrics
print("mean weights", mean_weights)
# Display computed maximum values
print("max weights", max_weights)
# Display computed minimum values
print("min weights", min_weights)
````

### Study Metrics Correlation
This cell implements a scatter plot comparing a list of student study hours against their final test outcomes using the seaborn visual framework.

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Test_Score": [50, 60, 65, 70, 75, 85]
}

df = pd.DataFrame(data)

sns.scatterplot(x="Hours_Studied", y="Test_Score", data=df)
plt.title("Relationship Between Study Hours and Test Scores")
plt.show()
````

````python
# Import the seaborn library for statistical charts
import seaborn as sns
# Import the pandas library for structured tables
import pandas as pd
# Import pyplot from matplotlib for overall plot configurations
import matplotlib.pyplot as plt

# Configure raw structured values mapping hours studied against final test grades
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Test_Score": [50, 60, 65, 70, 75, 85]
}

# Transform the raw structured dictionary into a standard pandas DataFrame
df = pd.DataFrame(data)

# Construct a scatter plot mapping hours studied on the X-axis and test scores on the Y-axis
sns.scatterplot(x="Hours_Studied", y="Test_Score", data=df)
# Set a descriptive title for the visual diagram
plt.title("Relationship Between Study Hours and Test Scores")
# Render the finalized figure window to the viewer
plt.show()
````
